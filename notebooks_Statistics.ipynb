````python
#Statistics 

#Generate a list of 100 integers containing values between 90 to 130 and store it in the variable `int_list`. After generating the list, find the following:


import random
import math
from collections import Counter

# Generate 100 integers between 90 and 130 (inclusive)
int_list = [random.randint(90, 130) for _ in range(100)]

print("Generated List:\n", int_list)


# (i) Calculate Mean
def calculate_mean(numbers):
    return sum(numbers) / len(numbers)


# (i) Calculate Median
def calculate_median(numbers):
    sorted_numbers = sorted(numbers)
    n = len(sorted_numbers)
    mid = n // 2

    if n % 2 == 0:
        return (sorted_numbers[mid - 1] + sorted_numbers[mid]) / 2
    else:
        return sorted_numbers[mid]


# (ii) Calculate Mode
def calculate_mode(numbers):
    counts = Counter(numbers)
    max_frequency = max(counts.values())
    return [num for num, freq in counts.items()
            if freq == max_frequency]


# (iii) Calculate Weighted Mean
def calculate_weighted_mean(values, weights):
    if len(values) != len(weights):
        raise ValueError("Values and weights must have equal lengths")
    if sum(weights) == 0:
        raise ValueError("Sum of weights cannot be zero")
    return sum(v * w for v, w in zip(values, weights)) / sum(weights)


# (iv) Calculate Geometric Mean
def calculate_geometric_mean(numbers):
    if not numbers or any(n <= 0 for n in numbers):
        raise ValueError("Numbers must be positive")
    return math.exp(sum(math.log(n) for n in numbers) / len(numbers))


# (v) Calculate Harmonic Mean
def calculate_harmonic_mean(numbers):
    if not numbers or any(n <= 0 for n in numbers):
        raise ValueError("Numbers must be positive")
    return len(numbers) / sum(1 / n for n in numbers)


# (vi) Calculate Midrange
def calculate_midrange(numbers):
    return (min(numbers) + max(numbers)) / 2


# (vii) Calculate Trimmed Mean
def calculate_trimmed_mean(numbers, percentage=10):
    if not numbers:
        raise ValueError("List cannot be empty")
    if not 0 <= percentage < 50:
        raise ValueError("Percentage must be between 0 and 50")

    sorted_numbers = sorted(numbers)
    trim_count = int(len(numbers) * percentage / 100)

    if trim_count == 0:
        trimmed_numbers = sorted_numbers
    else:
        trimmed_numbers = sorted_numbers[trim_count:-trim_count]

    return calculate_mean(trimmed_numbers)


# Display results
print("\n(i) Mean:", calculate_mean(int_list))
print("(i) Median:", calculate_median(int_list))
print("(ii) Mode:", calculate_mode(int_list))

# Example values and weights for weighted mean
weights = [1] * len(int_list)
print("(iii) Weighted Mean:", calculate_weighted_mean(int_list, weights))

print("(iv) Geometric Mean:", calculate_geometric_mean(int_list))
print("(v) Harmonic Mean:", calculate_harmonic_mean(int_list))
print("(vi) Midrange:", calculate_midrange(int_list))
print("(vii) Trimmed Mean (10%):", calculate_trimmed_mean(int_list, 10))

````

````python
#Generate a list of 500 integers containing values between 200 to 300 and store it in the variable `int_list2`. After generating the list, find the following:



import random
import math
import statistics
import matplotlib.pyplot as plt

# Generate 500 integers between 200 and 300 (inclusive)
int_list2 = [random.randint(200, 300) for _ in range(500)]

print("Generated list:", int_list2)
print("Total number of values:", len(int_list2))


# (i) Visualizations

# Calculate mean and standard deviation for Gaussian curve
mean_value = statistics.mean(int_list2)
std_dev = statistics.pstdev(int_list2)

# Frequency histogram
plt.figure(figsize=(8, 5))
plt.hist(int_list2, bins=20, density=True, alpha=0.6,
         edgecolor="black", label="Frequency distribution")

# Gaussian distribution curve
x_values = [
    min(int_list2) + i * (max(int_list2) - min(int_list2)) / 200
    for i in range(201)
]
gaussian_values = [
    (1 / (std_dev * math.sqrt(2 * math.pi))) *
    math.exp(-0.5 * ((x - mean_value) / std_dev) ** 2)
    for x in x_values
]
plt.plot(x_values, gaussian_values, label="Gaussian distribution")
plt.title("Frequency and Gaussian Distribution")
plt.xlabel("Values")
plt.ylabel("Density")
plt.legend()
plt.show()


# Smoothened KDE plot using a manual Gaussian kernel
def calculate_kde(data, points=200):
    n = len(data)
    data_mean = statistics.mean(data)
    data_std = statistics.pstdev(data)

    # Silverman's rule of thumb for bandwidth
    bandwidth = 1.06 * data_std * (n ** (-1 / 5))

    if bandwidth == 0:
        bandwidth = 1

    x_min = min(data) - 3 * bandwidth
    x_max = max(data) + 3 * bandwidth
    x_values = [
        x_min + i * (x_max - x_min) / (points - 1)
        for i in range(points)
    ]

    kde_values = []
    for x in x_values:
        density = sum(
            math.exp(-0.5 * ((x - value) / bandwidth) ** 2)
            for value in data
        )
        density /= n * bandwidth * math.sqrt(2 * math.pi)
        kde_values.append(density)

    return x_values, kde_values


kde_x, kde_y = calculate_kde(int_list2)

# (i.2) Frequency and smoothened KDE
plt.figure(figsize=(8, 5))
plt.hist(int_list2, bins=20, density=True, alpha=0.5,
         edgecolor="black", label="Frequency distribution")
plt.plot(kde_x, kde_y, label="Smoothed KDE")
plt.title("Frequency and Smoothened KDE Plot")
plt.xlabel("Values")
plt.ylabel("Density")
plt.legend()
plt.show()


# (i.3) Gaussian distribution and KDE
plt.figure(figsize=(8, 5))
plt.plot(x_values, gaussian_values, label="Gaussian distribution")
plt.plot(kde_x, kde_y, label="Smoothed KDE")
plt.title("Gaussian Distribution and Smoothened KDE")
plt.xlabel("Values")
plt.ylabel("Density")
plt.legend()
plt.show()


# (ii) Range
def calculate_range(numbers):
    return max(numbers) - min(numbers)


# (iii) Variance and standard deviation
def calculate_variance(numbers):
    mean = statistics.mean(numbers)
    return sum((x - mean) ** 2 for x in numbers) / len(numbers)


def calculate_standard_deviation(numbers):
    return math.sqrt(calculate_variance(numbers))


# (iv) Interquartile range (IQR)
def calculate_iqr(numbers):
    sorted_numbers = sorted(numbers)
    n = len(sorted_numbers)
    lower_half = sorted_numbers[:n // 2]
    upper_half = sorted_numbers[(n + 1) // 2:]

    q1 = statistics.median(lower_half)
    q3 = statistics.median(upper_half)

    return q3 - q1


# (v) Coefficient of variation
def calculate_cv(numbers):
    mean = statistics.mean(numbers)
    if mean == 0:
        raise ValueError("Mean cannot be zero")
    return (calculate_standard_deviation(numbers) / mean) * 100


# (vi) Mean absolute deviation (about the mean)
def calculate_mad(numbers):
    mean = statistics.mean(numbers)
    return sum(abs(x - mean) for x in numbers) / len(numbers)


# (vii) Quartile deviation (semi-interquartile range)
def calculate_quartile_deviation(numbers):
    return calculate_iqr(numbers) / 2


# (viii) Range-based coefficient of dispersion
def calculate_range_dispersion(numbers):
    denominator = max(numbers) + min(numbers)
    if denominator == 0:
        raise ValueError("Maximum + minimum cannot be zero")
    return (max(numbers) - min(numbers)) / denominator


# Display all statistical results
print("\n--- Statistical Results ---")
print("Range:", calculate_range(int_list2))
print("Variance (population):", calculate_variance(int_list2))
print("Standard deviation (population):",
      calculate_standard_deviation(int_list2))
print("Interquartile range (IQR):", calculate_iqr(int_list2))
print("Coefficient of variation (%):", calculate_cv(int_list2))
print("Mean absolute deviation (MAD):", calculate_mad(int_list2))
print("Quartile deviation:", calculate_quartile_deviation(int_list2))
print("Range-based coefficient of dispersion:",
      calculate_range_dispersion(int_list2))

````

````python
#Write a Python class representing a discrete random variable with methods to calculate its expected value and variance.


class DiscreteRandomVariable:
    def __init__(self, values, probabilities):
        if len(values) != len(probabilities):
            raise ValueError("Values and probabilities must have equal lengths")

        if any(p < 0 for p in probabilities):
            raise ValueError("Probabilities cannot be negative")

        if not math.isclose(sum(probabilities), 1.0):
            raise ValueError("Probabilities must sum to 1")

        self.values = values
        self.probabilities = probabilities

    # Calculate expected value
    def expected_value(self):
        return sum(x * p for x, p in zip(self.values, self.probabilities))

    # Calculate variance
    def variance(self):
        mean = self.expected_value()
        return sum(p * (x - mean) ** 2
                   for x, p in zip(self.values, self.probabilities))


# Example usage
import math

values = [1, 2, 3, 4, 5]
probabilities = [0.1, 0.2, 0.4, 0.2, 0.1]

rv = DiscreteRandomVariable(values, probabilities)

print("Expected Value:", rv.expected_value())
print("Variance:", rv.variance())
````

````python
#Write a Python function to calculate the probability density function (PDF) of a continuous random variable for a given normal distribution.


import math

def normal_pdf(x, mean, std_dev):
    if std_dev <= 0:
        raise ValueError("Standard deviation must be positive")

    pdf = (1 / (std_dev * math.sqrt(2 * math.pi))) * math.exp(
        -0.5 * ((x - mean) / std_dev) ** 2
    )
    return pdf


# Example usage
mean = 0
std_dev = 1
x = 1

result = normal_pdf(x, mean, std_dev)

print("PDF value:", result)
````

````python
#Create a program to calculate the cumulative distribution function (CDF) of exponential distribution.


import math

def exponential_cdf(x, rate):
    if rate <= 0:
        raise ValueError("Rate parameter must be positive")

    if x < 0:
        return 0.0

    return 1 - math.exp(-rate * x)


# Example usage
rate = 0.5
x = 2

result = exponential_cdf(x, rate)

print("CDF value:", result)
````

````python
#Write a Python function to calculate the probability mass function (PMF) of Poisson distribution.


import math

def poisson_pmf(k, lam):
    if k < 0 or not isinstance(k, int):
        raise ValueError("k must be a non-negative integer")
    if lam < 0:
        raise ValueError("Lambda must be non-negative")

    return (math.exp(-lam) * lam**k) / math.factorial(k)

# Example
k = 3
lam = 2

probability = poisson_pmf(k, lam)
print("Poisson PMF:", probability)
````

````python
#A company wants to test if a new website layout leads to a higher conversion rate (percentage of visitors who make a purchase). They collect data from the old and new layouts to compare.to generate the data use the following command:```python import numpy as np
# 50 purchases out of 1000 visitors old_layout = np.array([1] * 50 + [0] * 950)
# 70 purchases out of 1000 visitors  new_layout = np.array([1] * 70 + [0] * 930)```Apply z-test to find which layout is successful.


import numpy as np
import math

# Generate data
old_layout = np.array([1] * 50 + [0] * 950)
new_layout = np.array([1] * 70 + [0] * 930)

# Sample sizes and number of purchases
n1 = len(old_layout)
n2 = len(new_layout)

x1 = np.sum(old_layout)
x2 = np.sum(new_layout)

# Conversion rates
p1 = x1 / n1
p2 = x2 / n2

# Pooled proportion
p_pool = (x1 + x2) / (n1 + n2)

# Standard error under the null hypothesis
se = math.sqrt(p_pool * (1 - p_pool) * (1/n1 + 1/n2))

# Z-statistic
z = (p2 - p1) / se

# One-tailed p-value, testing whether new > old
p_value = 0.5 * math.erfc(z / math.sqrt(2))

print(f"Old layout conversion rate: {p1 * 100:.2f}%")
print(f"New layout conversion rate: {p2 * 100:.2f}%")
print(f"Z-statistic: {z:.4f}")
print(f"P-value: {p_value:.4f}")

if p_value < 0.05:
    print("The new layout is significantly more successful.")
else:
    print("Insufficient evidence that the new layout is more successful.")


    
````

````python
#A tutoring service claims that its program improves students' exam scores. A sample of students who participated in the program was taken, and their scores before and after the program were recorded.


import numpy as np
import math

# Generate samples
before_program = np.array([75, 80, 85, 70, 90, 78, 92, 88, 82, 87])
after_program = np.array([80, 85, 90, 80, 92, 80, 95, 90, 85, 88])

# Calculate differences
differences = after_program - before_program

# Calculate mean and standard deviation of differences
n = len(differences)
mean_diff = np.mean(differences)
std_diff = np.std(differences, ddof=1)

# Standard error and z-statistic
se = std_diff / math.sqrt(n)
z_stat = mean_diff / se

# One-tailed p-value: probability of improvement
p_value = 0.5 * math.erfc(z_stat / math.sqrt(2))

print("Mean score before:", np.mean(before_program))
print("Mean score after:", np.mean(after_program))
print("Mean improvement:", mean_diff)
print("Z-statistic:", round(z_stat, 4))
print("P-value:", round(p_value, 4))

if p_value < 0.05:
    print("The tutor's claim is supported: scores improved significantly.")
else:
    print("Insufficient evidence to support the tutor's claim.")
````

````python
#A company is testing two different website layouts to see which one leads to higher click-through rates. Write a Python function to perform an A/B test analysis, including calculating the t-statistic, degrees of freedom, and p-value.


import numpy as np
from scipy import stats

def ab_test(layout_a, layout_b):
    # Convert data to arrays
    a = np.array(layout_a)
    b = np.array(layout_b)

    # Perform Welch's independent two-sample t-test
    t_statistic, p_value = stats.ttest_ind(a, b, equal_var=False)

    # Calculate degrees of freedom (Welch-Satterthwaite formula)
    var_a = np.var(a, ddof=1)
    var_b = np.var(b, ddof=1)
    n_a = len(a)
    n_b = len(b)

    numerator = (var_a / n_a + var_b / n_b) ** 2
    denominator = ((var_a / n_a) ** 2 / (n_a - 1)
                   + (var_b / n_b) ** 2 / (n_b - 1))
    df = numerator / denominator

    print("Mean clicks - Layout A:", round(np.mean(a), 2))
    print("Mean clicks - Layout B:", round(np.mean(b), 2))
    print("T-statistic:", round(t_statistic, 4))
    print("Degrees of freedom:", round(df, 2))
    print("P-value:", f"{p_value:.8f}")

    if p_value < 0.05:
        print("There is a significant difference between the layouts.")
        if np.mean(b) > np.mean(a):
            print("Layout B has the higher average clicks.")
        else:
            print("Layout A has the higher average clicks.")
    else:
        print("No statistically significant difference was found.")

# Given data
layout_a_clicks = [28, 32, 33, 29, 31, 34, 30, 35, 36, 37]
layout_b_clicks = [40, 41, 38, 42, 39, 44, 43, 41, 45, 47]

# Run A/B test
ab_test(layout_a_clicks, layout_b_clicks)
````

````python
#A pharmaceutical company wants to determine if a new drug is more effective than an existing drug in reducing cholesterol levels. Create a program to analyze the clinical trial data and calculate the statistic and p-value for the treatment effect.


import numpy as np
from scipy import stats

def analyze_drugs(existing_drug_levels, new_drug_levels):
    existing = np.array(existing_drug_levels)
    new = np.array(new_drug_levels)

    # Calculate mean cholesterol levels
    mean_existing = np.mean(existing)
    mean_new = np.mean(new)

    # Welch's independent two-sample t-test
    t_stat, p_value = stats.ttest_ind(
        existing, new, equal_var=False, alternative='greater'
    )

    # Calculate degrees of freedom
    var_existing = np.var(existing, ddof=1)
    var_new = np.var(new, ddof=1)
    n1 = len(existing)
    n2 = len(new)

    numerator = (var_existing/n1 + var_new/n2)**2
    denominator = (
        (var_existing/n1)**2/(n1-1)
        + (var_new/n2)**2/(n2-1)
    )
    df = numerator / denominator

    print("Mean cholesterol (existing drug):", round(mean_existing, 2))
    print("Mean cholesterol (new drug):", round(mean_new, 2))
    print("T-statistic:", round(t_stat, 4))
    print("Degrees of freedom:", round(df, 2))
    print("P-value:", f"{p_value:.6f}")

    if p_value < 0.05:
        print("The new drug significantly reduces cholesterol.")
    else:
        print("Insufficient evidence that the new drug is more effective.")

# Given data
existing_drug_levels = [180, 182, 175, 185, 178, 176, 172, 184, 179, 183]
new_drug_levels = [170, 172, 165, 168, 175, 173, 170, 178, 172, 176]

analyze_drugs(existing_drug_levels, new_drug_levels)
````

````python
#A school district introduces an educational intervention program to improve math scores. Write a Python function to analyze pre- and post-intervention test scores, calculating the t-statistic and p-value to determine if the intervention had a significant impact. 


import numpy as np
from scipy import stats

def analyze_intervention(pre_scores, post_scores):
    # Convert scores to arrays
    pre = np.array(pre_scores)
    post = np.array(post_scores)

    # Paired t-test
    t_statistic, p_value = stats.ttest_rel(post, pre)

    # Calculate mean scores
    pre_mean = np.mean(pre)
    post_mean = np.mean(post)

    print("Mean score before intervention:", round(pre_mean, 2))
    print("Mean score after intervention:", round(post_mean, 2))
    print("Mean improvement:", round(post_mean - pre_mean, 2))
    print("T-statistic:", round(t_statistic, 4))
    print("P-value:", f"{p_value:.6f}")

    # Decision at 5% significance level
    if p_value < 0.05:
        print("The intervention had a statistically significant impact.")
        if post_mean > pre_mean:
            print("The intervention significantly improved test scores.")
    else:
        print("No statistically significant impact was found.")

# Given data
pre_intervention_scores = [80, 85, 90, 75, 88, 82, 92, 78, 85, 87]
post_intervention_scores = [90, 92, 88, 92, 95, 91, 96, 93, 89, 93]

# Run the analysis
analyze_intervention(pre_intervention_scores, post_intervention_scores)
````

````python
#An HR department wants to investigate if there's a gender-based salary gap within the company. Develop a program to analyze salary data, calculate the t-statistic, and determine if there's a statistically significant difference between the average salaries of male and female employees.


import numpy as np
from scipy import stats

def analyze_salary_gap(male_salaries, female_salaries):
    male = np.array(male_salaries)
    female = np.array(female_salaries)

    # Calculate average salaries
    male_mean = np.mean(male)
    female_mean = np.mean(female)

    # Welch's two-sample t-test (two-tailed)
    t_stat, p_value = stats.ttest_ind(
        male, female, equal_var=False
    )

    print("Average male salary:", round(male_mean, 2))
    print("Average female salary:", round(female_mean, 2))
    print("Salary difference:", round(male_mean - female_mean, 2))
    print("T-statistic:", round(t_stat, 4))
    print("P-value:", round(p_value, 4))

    if p_value < 0.05:
        print("There is a statistically significant salary difference.")
    else:
        print("No statistically significant salary difference was found.")

# Generate synthetic salary data
np.random.seed(0)
male_salaries = np.random.normal(loc=50000, scale=10000, size=20)
female_salaries = np.random.normal(loc=55000, scale=9000, size=20)

# Run analysis
analyze_salary_gap(male_salaries, female_salaries)
````

````python
#A manufacturer produces two different versions of a product and wants to compare their quality scores. Create a Python function to analyze quality assessment data, calculate the t-statistic, and decide whether there's a significant difference in quality between the two versions.


import numpy as np
from scipy import stats

def analyze_quality(version1_scores, version2_scores):
    v1 = np.array(version1_scores)
    v2 = np.array(version2_scores)

    # Calculate average quality scores
    mean1 = np.mean(v1)
    mean2 = np.mean(v2)

    # Welch's independent two-sample t-test
    t_stat, p_value = stats.ttest_ind(
        v1, v2, equal_var=False
    )

    print("Mean quality score - Version 1:", round(mean1, 2))
    print("Mean quality score - Version 2:", round(mean2, 2))
    print("Difference in means:", round(mean1 - mean2, 2))
    print("T-statistic:", round(t_stat, 4))
    print("P-value:", f"{p_value:.10f}")

    # Decision at 5% significance level
    if p_value < 0.05:
        print("There is a significant difference in quality.")
        if mean1 > mean2:
            print("Version 1 has the higher average quality score.")
        else:
            print("Version 2 has the higher average quality score.")
    else:
        print("No statistically significant difference was found.")

# Given data
version1_scores = [85, 88, 82, 89, 87, 84, 90, 88, 85, 86,
                   91, 83, 87, 84, 89, 86, 84, 88, 85, 86,
                   89, 90, 87, 88, 85]

version2_scores = [80, 78, 83, 81, 79, 82, 76, 80, 78, 81,
                   77, 82, 80, 79, 82, 79, 80, 81, 79, 82,
                   79, 78, 80, 81, 82]

# Run the analysis
analyze_quality(version1_scores, version2_scores)
````

````python
# A restaurant chain collects customer satisfaction scores for two different branches. Write a program to analyze the scores, calculate the t-statistic, and determine if there's a statistically significant difference in customer satisfaction between the branches.


import numpy as np
import math

def analyze_satisfaction(branch_a_scores, branch_b_scores):
    a = np.array(branch_a_scores, dtype=float)
    b = np.array(branch_b_scores, dtype=float)

    # Calculate means and sample variances
    mean_a = np.mean(a)
    mean_b = np.mean(b)
    var_a = np.var(a, ddof=1)
    var_b = np.var(b, ddof=1)
    n_a = len(a)
    n_b = len(b)

    # Welch's t-statistic
    se_squared = var_a / n_a + var_b / n_b
    t_stat = (mean_a - mean_b) / math.sqrt(se_squared)

    # Welch-Satterthwaite degrees of freedom
    df = se_squared**2 / (
        (var_a / n_a)**2 / (n_a - 1)
        + (var_b / n_b)**2 / (n_b - 1)
    )

    # Two-tailed p-value using a normal approximation
    p_value = math.erfc(abs(t_stat) / math.sqrt(2))

    print("Mean score - Branch A:", round(mean_a, 4))
    print("Mean score - Branch B:", round(mean_b, 4))
    print("T-statistic:", round(t_stat, 4))
    print("Degrees of freedom:", round(df, 4))
    print("Approximate p-value:", f"{p_value:.6f}")

    if p_value < 0.05:
        print("There is a statistically significant difference.")
    else:
        print("No statistically significant difference was found.")

# Given data
branch_a_scores = [4, 5, 3, 4, 5, 4, 5, 3, 4, 4, 5, 4, 4, 3, 4,
                   5, 5, 4, 3, 4, 5, 4, 3, 5, 4, 4, 5, 3, 4, 5, 4]

branch_b_scores = [3, 4, 2, 3, 4, 3, 4, 2, 3, 3, 4, 3, 3, 2, 3,
                   4, 4, 3, 2, 3, 4, 3, 2, 4, 3, 3, 4, 2, 3, 4, 3]

analyze_satisfaction(branch_a_scores, branch_b_scores)
````

````python
#A political analyst wants to determine if there is a significant association between age groups and voter preferences (Candidate A or Candidate B). They collect data from a sample of 500 voters and classify them into different age groups and candidate preferences. Perform a Chi-Square test to determine if there is a significant association between age groups and voter preferences.


import numpy as np
from scipy.stats import chi2_contingency

# Generate data
np.random.seed(0)
age_groups = np.random.choice(['18-30', '31-50', '51+', '51+'], size=30)
voter_preferences = np.random.choice(
    ['Candidate A', 'Candidate B'], size=30
)

# Create contingency table
age_categories = ['18-30', '31-50', '51+']
candidates = ['Candidate A', 'Candidate B']

observed = np.zeros((len(age_categories), len(candidates)), dtype=int)

for age, preference in zip(age_groups, voter_preferences):
    i = age_categories.index(age)
    j = candidates.index(preference)
    observed[i, j] += 1

# Perform Chi-Square test
chi2, p_value, dof, expected = chi2_contingency(observed)

print("Observed frequency table:")
print(observed)
print("\nChi-Square statistic:", round(chi2, 4))
print("Degrees of freedom:", dof)
print("P-value:", round(p_value, 4))

if p_value < 0.05:
    print("There is a significant association between age group and voter preference.")
else:
    print("No statistically significant association was found.")
````

````python
#A company conducted a customer satisfaction survey to determine if there is a significant relationship between product satisfaction levels (Satisfied, Neutral, Dissatisfied) and the region where customers are located (East, West, North, South). The survey data is summarized in a contingency table. Conduct a ChiSquare test to determine if there is a significant relationship between product satisfaction levels and customer regions.


import numpy as np
from scipy.stats import chi2_contingency

# Sample data: Satisfaction levels (rows) vs. Regions (columns)
data = np.array([
    [50, 30, 40, 20],
    [30, 40, 30, 50],
    [20, 30, 40, 30]
])

# Perform Chi-Square test
chi2, p_value, dof, expected = chi2_contingency(data)

print("Observed Data:")
print(data)

print("\nExpected Frequencies:")
print(np.round(expected, 2))

print("\nChi-Square Statistic:", round(chi2, 4))
print("Degrees of Freedom:", dof)
print("P-value:", round(p_value, 4))

# Decision at 5% significance level
if p_value < 0.05:
    print("\nThere is a significant relationship between satisfaction and region.")
else:
    print("\nNo statistically significant relationship was found.")
````

````python
#A company implemented an employee training program to improve job performance (Effective, Neutral, Ineffective). After the training, they collected data from a sample of employees and classified them based on their job performance before and after the training. Perform a Chi-Square test to determine if there is a significant difference between job performance levels before and after the training.


import numpy as np
from scipy.stats import chi2_contingency

# Rows: performance before training
# Columns: performance after training
# Categories: Effective, Neutral, Ineffective
data = np.array([
    [50, 30, 20],
    [30, 40, 30],
    [20, 30, 40]
])

chi2, p_value, dof, expected = chi2_contingency(data)

print("Chi-Square Statistic:", round(chi2, 4))
print("Degrees of Freedom:", dof)
print("P-value:", round(p_value, 6))
print("\nExpected Frequencies:")
print(expected)

alpha = 0.05

if p_value < alpha:
    print("\nConclusion: There is a significant association between")
    print("performance classification before and after training.")
else:
    print("\nConclusion: There is no significant association between")
    print("performance classification before and after training.")
````

````python
#A company produces three different versions of a product: Standard, Premium, and Deluxe. The company wants to determine if there is a significant difference in customer satisfaction scores among the three product versions. They conducted a survey and collected customer satisfaction scores for each version from a random sample of customers. Perform an ANOVA test to determine if there is a significant difference in customer satisfaction scores.


import numpy as np
from scipy.stats import f_oneway

# Customer satisfaction scores
standard_scores = [80, 85, 90, 78, 88, 82, 92, 78, 85, 87]

premium_scores = [90, 92, 88, 92, 95, 91, 96, 93, 89, 93]

deluxe_scores = [95, 98, 92, 97, 96, 94, 98, 97, 92, 99]

# Calculate group means
print("Standard Mean:", np.mean(standard_scores))
print("Premium Mean:", np.mean(premium_scores))
print("Deluxe Mean:", np.mean(deluxe_scores))

# Perform one-way ANOVA
f_statistic, p_value = f_oneway(
    standard_scores,
    premium_scores,
    deluxe_scores
)

print("\nF-statistic:", round(f_statistic, 4))
print("P-value:", p_value)

# Interpret the result
if p_value < 0.05:
    print("Conclusion: Significant difference in satisfaction scores.")
else:
    print("Conclusion: No significant difference in satisfaction scores.")
````
